4. A projectile is launched with initial velocity u = 20 m s−1 at an angle θ = 45◦.
(a) Plot x(t) and y(t) as functions of time.
(b) Plot the trajectory y versus x.

```python
#Author: Ataul Islam [PHYS064]

#A projectile is launched with initial velocity u = 20 m s−1 at an angle θ = 45◦.
#(a) Plot x(t) and y(t) as functions of time.
#(b) Plot the trajectory y versus x.

import matplotlib.pyplot as plt
import numpy as np

u = 20.0 #m/s
theta = 45 # in degrees
theta = theta * np.pi/ 180
g = 9.8 # m/s^2
t_f = 2 * u * np.sin(theta) / g
t = np.linspace(0,t_f,100)
x = u*np.cos(theta)*t
y = u*np.sin(theta)*t - 0.5*g*t**2

fig = plt.figure(figsize = (10,8))

#plotting first graph
ax1 = plt.subplot(2,2,1)
ax1.plot(t,x, color = "red", lw = 2, label = "Horizontal Distance")
ax1.set_title("Plot of x(t) vs t in Projectile Motion")
ax1.set_xlabel("Time")
ax1.set_ylabel("Horizontal Distance Travelled (x(t))")
ax1.legend()
ax1.grid()

#plotting second graph
ax2 = plt.subplot(2,2,2)
ax2.plot(t,y, color = "green", lw = 2,  label = "Vertical Distance")
ax2.set_title("Plot of y(t) vs t in Projectile Motion")
ax2.set_xlabel("Time")
ax2.set_ylabel("Vertical Distance Travelled (y(t))")
ax2.legend()
ax2.grid()
#plotting third graph
ax3 = plt.subplot(2,2,(3,4))
ax3.plot(x,y, color = "blue", lw = 2, ls = "--", label = "Trajectory of Projectile")
ax3.set_title("Plot of y vs x: Trajectory of Projectile")
ax3.set_xlabel("Increasing X")
ax3.set_ylabel("Increasing Y")
ax3.legend()
ax3.grid()


plt.tight_layout()
plt.show()
```

5. The displacement of a particle executing SHM is
x(t) = Asin(ωt),
where A = 2 m and ω = 3 rad s−1.
(a) Plot displacement versus time.
(b) Plot velocity and acceleration versus time on the same graph.

```python
#Author: Ataul Islam [PHYS064]

#5. The displacement of a particle executing SHM is
#x(t) = Asin(ωt),
#where A = 2 m and ω = 3 rad s−1.
#(a) Plot displacement versus time.
#(b) Plot velocity and acceleration versus time on the same graph.

import matplotlib.pyplot as plt
import numpy as np

A = 2.0 #in m
w = 3.0 #in rad s-1
t = np.linspace(0,10,500)
x = A * np.sin(w * t)
v = A * w * np.cos(w * t)
a = - A * w * w * np.sin(w * t)

fig = plt.figure(figsize = (10,8))

ax1 = plt.subplot(2,2,(1,2))
ax1.plot(t,x, color = "red", lw = 2, label = "Displacement")
ax1.set_title("Displacement versus Time Graph")
ax1.set_xlabel("Time")
ax1.set_ylabel("Displacement")
ax1.grid()
ax1.legend()

ax2 = plt.subplot(2,2,(3,4))
ax2.plot(t, v, color = "green", lw = 2, label = "Velocity")
ax2.plot(t, a, color = "blue", lw = 2, ls = "--", label = "Acceleration")
ax2.set_title("Velocity & Acceleration versus Time Graph")
ax2.set_xlabel("Time")
ax2.set_ylabel("Velocity & Acceleration")
ax2.legend()
ax2.grid()



plt.tight_layout()
plt.show()
```

6. The displacement of a damped oscillator is
x(t) = Ae−βt cos(ωt),
with A = 1, β = 0.2, and ω = 2.
(a) Plot x(t) versus t.
(b) Compare the plot with an undamped oscillator.

```python
#Author: Ataul Islam [PHYS064]

#6. The displacement of a damped oscillator is
#x(t) = Ae−βt cos(ωt),
#with A = 1, β = 0.2, and ω = 2.
#(a) Plot x(t) versus t.
#(b) Compare the plot with an undamped oscillator.

import matplotlib.pyplot as plt
import numpy as np

t = np.linspace(0,10,500)
A = 1.0
B = 0.2 #damping constant
w = 2.0

x = A * np.exp(- B * t) * np.cos(w * t)
y = A * np.cos(w * t)

fig = plt.figure(figsize = (10,8))

ax1 = plt.subplot(2,2,(1,2))
ax1.plot(t, x, color = "red", lw = 2, label = "Damped Oscillator")
ax1.set_title("Displacement of a Damped Oscillator")
ax1.set_xlabel("Time")
ax1.set_ylabel("Displacement")
ax1.legend()
ax1.grid()


ax2 = plt.subplot(2,2,(3,4))
ax2.plot(t, x, color = "red", lw = 2, label = "Damped Oscillator")
ax2.plot(t, y, color = "blue", lw = 2, label = "Undamped Oscillator")
ax2.set_title("Comparison of Damped and Undamped Oscillator")
ax2.set_xlabel("Time")
ax2.set_ylabel("Displacement")
ax2.legend()
ax2.grid()

plt.tight_layout()
plt.show()
```

```python
#Author : Ataul Islam [PHYS064]

import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-5,5, 500)

def pdf(x, u , o):
    f = 1/(o * np.sqrt(2 * np.pi))* np.exp( - ((x - u)** 2)/(2 * o ** 2))
    return f

fig = plt.figure(figsize = (10, 8))

ax1 = plt.subplot(2,2,(1,2))
ax1.plot(x, pdf(x, 0, 1), color = "blue", lw = 2)
ax1.set_title("Normal Distribution for  μ = 0, σ = 1")
ax1.set_xlabel("Changing Value of X")
ax1.set_ylabel("Probability Density Function")
ax1.grid()

ax2 = plt.subplot(2,2, (3,4))
ax2.plot(x, pdf(x, 0, 1), color = "blue", lw = 2, label = "μ = 0, σ = 1")
ax2.plot(x, pdf(x, 0, 0.5), color = "red", lw = 2, label = "μ = 0, σ = 0.5")
ax2.plot(x, pdf(x, 0, 2), color = "green", lw = 2, label = " μ= 0, σ = 2")
ax2.set_title("Different Normal Distributions")
ax2.set_xlabel("Changing Value of X")
ax2.set_ylabel("Probability Density Function")
ax2.legend()
ax2.grid()

plt.tight_layout()
plt.show()
```
